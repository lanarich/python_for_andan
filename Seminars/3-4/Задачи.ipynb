### Задание №5: написать программу, которая с помощью цикла while определяет, сколько гласных/согласных из перечисленных в подстроке

```python
def identifier(stroka):
    vowels = 'aeiouy'
    vowels_counter = 0
    notvowels_counter = 0
    i = 0
    while i < len(stroka):
        if stroka[i].lower() in vowels:
            vowels_counter += 1
        else:
            notvowels_counter += 1
        i += 1
    return vowels_counter, notvowels_counter

identifier('hello')
```
