 # Задачи

### Задание №1

Напишите программу, которая запрашивает радиус круга у пользователя и выводит сообщение:
"Площадь круга равна <значение>", где <значение> — результат вычисления πr².

### Задание №2  

Напишите программу, которая меняет значения двух переменных x и y местами без использования множественного присваивания (x, y = y, x).

### Задание №3

Напишите программу, которая меняет значения двух переменных x и y местами без использования множественного присваивания (x, y = y, x).



### Задание №4

Напишите программу, которая с помощью цикла for вычисляет сумму первых n натуральных чисел (от 1 до n включительно).

### Задание №5

Напишите программу, которая с помощью цикла while по введённой строке подсчитывает количество гласных и согласных, используя введённые пользователем наборы гласных и согласных символов. Выведите: Гласные: X, Согласные: Y.

### Задание №6

Напишите программу, которая проверяет, является ли введённое число квадратом целого числа.

### Задание №7

Доработайте код предыдущего задания так, чтобы программа проверяла, является ли число кубом целого числа.

### Задание №8

Напишите программу, которая принимает количество секунд и преобразует его в формат: "дни:часы:минуты:секунды".

### Задание №9

Напишите программу, которая вычисляет сумму всех цифр введённого целого числа.

### Задание №10

При заданном целом числе n вычислите выражение n + nn + nnn. Например, если n = 3, результат должен быть 3 + 33 + 333 = 369.

### Задание №11

Для заданного числа найдите и выведите цифру, которая встречается чаще всего. Если таких цифр несколько — выведите любую из них.

### Задание №12

Для заданного числа найдите и выведите цифру, которая встречается реже всего. Если таких цифр несколько — выведите любую из них.

### Задание №13

Для двух заданных чисел выведите цифры, которые встречаются в записи обоих чисел (без повторов, в порядке возрастания).
